Exercise 2 — Code Generation with ReACT Prompting

```python
numbers=[10,20,30,40,50]
print("PLAN: Create a Python program that calculates the average of a list of numbers.")
print("ACTION: Generate and run the code.")
def calculate_average(numbers):
    try:
        if len(numbers)==0:
            return "Error: The list cannot be empty."
        return sum(numbers)/len(numbers)
    except TypeError:
        return "Error: The list must contain only numbers."
result=calculate_average(numbers)
print("OBSERVATION:",result)
print("REFLECTION: The code worked, but it should also handle an empty list and invalid data.")
test_empty=[]
test_invalid=[10,20,"hello",40]
print("FIX TEST - EMPTY LIST:",calculate_average(test_empty))
print("FIX TEST - INVALID DATA:",calculate_average(test_invalid))
print("FINAL RESULT:",calculate_average(numbers))
```
