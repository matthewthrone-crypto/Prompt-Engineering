Exercise 3 — Self-Reflection Prompt for Improving Output

```python
original_summary="Artificial intelligence helps businesses by automating tasks and analyzing data. It can save time and help companies make decisions."
print("ORIGINAL SUMMARY:",original_summary)
criteria=["Accuracy","Clarity","Audience","Length","Formatting"]
def self_reflect(summary):
    critique="The summary is accurate but too general. It should be clearer, include a specific example, stay under 75 words, and use simple language for a high school or early college audience."
    return critique
critique=self_reflect(original_summary)
print("SELF-CRITIQUE:",critique)
def improve_summary(summary,critique):
    improved="Artificial intelligence helps businesses save time by automating repetitive tasks and analyzing large amounts of data. For example, a company can use AI to answer customer questions or predict future sales. This helps employees work faster and make better decisions while still allowing people to review important results."
    return improved
improved_summary=improve_summary(original_summary,critique)
print("REQUIREMENTS:",criteria)
print("IMPROVED SUMMARY:",improved_summary)
```
