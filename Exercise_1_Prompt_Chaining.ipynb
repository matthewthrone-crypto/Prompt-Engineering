Exercise 1 — Prompt Chaining for a Customer Support AI

```python
customer_message="I ordered a laptop 10 days ago, but it still has not arrived."
def classify_issue(message):
    if "arrived" in message.lower() or "order" in message.lower():
        return "Shipping"
    return "Other"
issue=classify_issue(customer_message)
def gather_information(message,issue):
    if issue=="Shipping":
        return "Please provide your order number so I can check the shipping status."
    return "Please provide more information about the issue."
missing_info=gather_information(customer_message,issue)
def propose_solution(message,issue,missing_info):
    if issue=="Shipping":
        return "Once the order number is provided, check the tracking information. If the package is delayed, give the customer an updated delivery estimate."
    return "Review the customer's information and provide the appropriate support."
solution=propose_solution(customer_message,issue,missing_info)
def escalation_check(issue,solution):
    if issue in ["Billing","Return"]:
        return "YES - A human agent may need to review the customer's request."
    return "NO - The issue can currently be handled through the normal support process."
escalation=escalation_check(issue,solution)
print("CUSTOMER MESSAGE:",customer_message)
print("ISSUE CLASSIFICATION:",issue)
print("MISSING INFORMATION:",missing_info)
print("PROPOSED SOLUTION:",solution)
print("ESCALATION DECISION:",escalation)
```
